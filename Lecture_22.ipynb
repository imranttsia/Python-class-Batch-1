The Frozenset

-Immutable

-Unordered

-Hashable

-Unique Elements

```python
my_set={2,3,4,5,7}
print(type(my_set))
set_1=frozenset(my_set)
print(type(set_1))
```

```python
set_1.remove(3)
print(set_1)
```

```python

set_1.discard(3)
print(set_1)
```

```python
set_1.pop()
print(set_1)
```

```python
set_1.add(3)
print(set_1)
```

```python
set_1 = frozenset({2,3,4,5,7})
set_2 = frozenset({1,2,3,8})

print(f"Original frozenset_1: {set_1}")
print(f"Original frozenset_2: {set_2}")

# issubset()
print(f"Is set_2 a subset of set_1? {set_2.issubset(set_1)}")

# issuperset()
print(f"Is set_1 a superset of set_2? {set_1.issuperset(set_2)}")

# isdisjoint()
print(f"Are set_1 and set_2 disjoint? {set_1.isdisjoint(set_2)}")

# union()
union_set = set_1.union(set_2)
print(f"Union of set_1 and set_2: {union_set}")

# intersection()
intersection_set = set_1.intersection(set_2)
print(f"Intersection of set_1 and set_2: {intersection_set}")

# difference()
difference_set = set_1.difference(set_2)
print(f"Difference of set_1 and set_2 (set_1 - set_2): {difference_set}")

# symmetric_difference()
symmetric_difference_set = set_1.symmetric_difference(set_2)
print(f"Symmetric difference of set_1 and set_2: {symmetric_difference_set}")
```

```python
fhand = open('mbox.txt')
for line in fhand:
    if line.startswith('From:'):
        print(line)
```

```python
fhand = open('mbox.txt')
for line in fhand:
    line = line.rstrip()
    if line.startswith('From:'):
        print(line)
```

```python
fhand = open('mbox.txt')
for line in fhand:
    line = line.rstrip()
    # Skip 'uninteresting lines'
    if not line.startswith('From:'):
        continue
    # Process our 'interesting' line
    print(line)
```
