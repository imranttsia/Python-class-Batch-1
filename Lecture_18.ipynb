### Use Cases for `list.sort()` Method

The `sort()` method sorts the items of a list in-place. The changes are made directly to the original list, and it doesn't return a new list.

```python
# 1. Sorting a list of numbers in ascending order (default)
numbers = [3, 1, 4, 1, 5, 9, 2, 6]
print(f"Original numbers: {numbers}")
numbers.sort()
print(f"Sorted ascending: {numbers}")

# 2. Sorting a list of numbers in descending order
numbers_desc = [3, 1, 4, 1, 5, 9, 2, 6]
print(f"\nOriginal numbers (for descending): {numbers_desc}")
numbers_desc.sort(reverse=True)
print(f"Sorted descending: {numbers_desc}")
```

```python
# 3. Sorting a list of strings alphabetically
words = ["apple", "zebra", "banana", "cat", "dog"]
print(f"Original words: {words}")
words.sort()
print(f"Sorted alphabetically: {words}")

# 4. Sorting a list of strings by length (using a key function)
words_by_length = ["apple", "zebra", "banana", "cat", "dog"]
print(f"\nOriginal words (for sorting by length): {words_by_length}")
words_by_length.sort(key=len)
print(f"Sorted by length: {words_by_length}")
```

### Use Cases for `list.reverse()` Method

The `reverse()` method reverses the elements of a list *in-place*. This means it modifies the original list and does not return a new list.

```python
# 1. Reversing a list of numbers
numbers = [1, 2, 3, 4, 5]
print(f"Original numbers: {numbers}")
numbers.reverse()
print(f"Reversed numbers: {numbers}")
```

```python
# 2. Reversing a list of strings
words = ["apple", "banana", "cherry", "date"]
print(f"\nOriginal words: {words}")
words.reverse()
print(f"Reversed words: {words}")
```

```python
# 3. Reversing a list with mixed data types
mixed_list = [1, "hello", True, 3.14]
print(f"\nOriginal mixed list: {mixed_list}")
mixed_list.reverse()
print(f"Reversed mixed list: {mixed_list}")
```

```python
# 4. Chaining operations (reverse then print)
my_list = ['a', 'b', 'c']
print(f"\nOriginal list for chaining: {my_list}")
my_list.reverse() # Modifies in-place
print(f"After reverse: {my_list}")

# Note: you cannot directly do print(my_list.reverse()) because reverse() returns None
```

### Use Cases for `list.count()` Method

The `count()` method returns the number of times a specified value appears in the list.

```python
# 1. Counting occurrences of an integer
numbers = [1, 2, 2, 3, 4, 2, 5, 1, 6]
count_of_2 = numbers.count(2)
print(f"Original list: {numbers}")
print(f"Number of times 2 appears: {count_of_2}")
```

```python
# 2. Counting occurrences of a string
fruits = ["apple", "banana", "orange", "apple", "grape", "apple"]
count_of_apple = fruits.count("apple")
print(f"\nOriginal list: {fruits}")
print(f"Number of times 'apple' appears: {count_of_apple}")
```

```python
# 3. Counting occurrences of an element not in the list
colors = ["red", "green", "blue"]
count_of_yellow = colors.count("yellow")
print(f"\nOriginal list: {colors}")
print(f"Number of times 'yellow' appears: {count_of_yellow}")
```

```python
# 4. Counting occurrences of boolean values
boolean_list = [True, False, True, True, False]
count_of_true = boolean_list.count(True)
count_of_false = boolean_list.count(False)
print(f"\nOriginal list: {boolean_list}")
print(f"Number of times True appears: {count_of_true}")
print(f"Number of times False appears: {count_of_false}")
```

```python
# 5. Counting occurrences in a list of mixed data types
mixed_data = [1, 'hello', 2, 'world', 1, True, 'hello', 3.14]
count_of_1 = mixed_data.count(1)
count_of_hello = mixed_data.count('hello')
print(f"\nOriginal list: {mixed_data}")
print(f"Number of times 1 appears: {count_of_1}")
print(f"Number of times 'hello' appears: {count_of_hello}")
```

### Use Cases for `list.append()` Method

The `append()` method adds a single item to the end of a list. It modifies the list in-place and does not return any value (it returns `None`).

```python
# 1. Appending a single element (number)
my_list = [10, 20, 30]
print(f"Original list: {my_list}")
my_list.append(40)
print(f"After appending 40: {my_list}")
```

```python
# 2. Appending a string
names = ["Alice", "Bob"]
print(f"\nOriginal names: {names}")
names.append("Charlie")
print(f"After appending 'Charlie': {names}")
```

```python
# 3. Appending a list (as a single element)
list_of_lists = [[1, 2], [3, 4]]
print(f"\nOriginal list of lists: {list_of_lists}")
list_of_lists.append([5, 6])
print(f"After appending a new list: {list_of_lists}")

# Note: append treats the appended list as one single item
```

```python
# 5. Appending in a loop
squares = []
for i in range(5):
    squares.append(i * i)
print(f"\nList of squares: {squares}")
```

### Use Cases for `list.extend()` Method

The `extend()` method adds all the elements of an iterable (e.g., another list, tuple, string, or range) to the end of the current list. It modifies the list in-place and does not return a new list.

```python
# 1. Extending with another list
list1 = [1, 2, 3]
list2 = [4, 5, 6]
print(f"Original list1: {list1}")
list1.extend(list2)
print(f"After extending list1 with list2: {list1}")
```

```python
# 3. Extending with a string (iterates over characters)
word_list = ['h', 'e', 'l']
my_string = "lo"
print(f"\nOriginal word_list: {word_list}")
word_list.extend(my_string)
print(f"After extending with a string: {word_list}")
```

```python
# 4. Extending with a range object
num_list = [10, 20]
my_range = range(30, 60, 10) # Generates 30, 40, 50
print(f"\nOriginal num_list: {num_list}")
num_list.extend(my_range)
print(f"After extending with a range: {num_list}")
```

```python
# 5. Combining extend with a loop (less common, but possible)
combined_list = []
list_of_iterables = [[1, 2], (3, 4), 'five']
print(f"\nOriginal combined_list: {combined_list}")
for iterable in list_of_iterables:
    combined_list.extend(iterable)
print(f"After extending with multiple iterables in a loop: {combined_list}")
```
