break statement

```python
count=0
while count<5:
  count+=1
  if count==3:
    break
print(count)
```

continue statement

```python
count=0
while count<5:
  count+=1
  if count==3:
    continue
  print(count)
```

Nested Loops

```python
x=1
while x<=3:
  y=1
  while y<=2:
    print(x,y)
    y+=1
  x+=1

#
```

For loop

For loop (list)

```python
Person=["Imran", "Ayaan", "Zehra"]
for name in Person:
  print("Best Wishes: ",name)
```

```python
number=[1,2,5,6,8,12]
seq=0
for num in number:
  seq=num*num
  print(seq)
```

For loop (string)

```python
Person="Ayaan"
for character in Person:
  print(character)
```

```python
Person="Ayaan"
for character in Person:
  print(character,end="")
```

Loop (Tuple)

```python
t=(2,3,6,7,9)
for i in t:
  print(i)
```

For loop(Dictionary)

```python
student={"name":"Ayaan","age":16,"class":10,"section":"A"}
for key in student:
  print(key)
```

```python
student={"name":"Ayaan","age":16,"class":10,"section":"A"}
for item in student:
  print(item)
```

```python
student={"name":"Ayaan","age":16,"class":10,"section":"A"}
for item in student.values():
  print(item)
```

```python
student={"name":"Ayaan","age":16,"class":10,"section":"A"}
for key, value in student.items():
  print(f"Key: {key}, Value: {value}")
```

Examples

```python
total=0
for item in [3,5,8,9,12,34]:
  total+=item
print(item)
```

```python
my_list = [10, 20, 30, 40, 50]
total_sum = 0

for number in my_list:
  total_sum += number

print(f"The sum of all items in the list is: {total_sum}")
```

```python
largest=None
for num in [3,5,8,9,12,34]:
  if largest is None:
    largest=num
  elif num>largest:
    largest=num
print(largest)
```

Nested For loop

```python
for i in range(1,6):
  for j in range(0,i):
    print(i, end="")
  print("")
```
