Dictionary comprehension

```python
original_dict={"a":1,"b":2,"c":3}
print("Original_dict= ", original_dict)
new_dict={key:value**2 for key,value in original_dict.items()}
print("New_dict= ", new_dict)
```

```python
original_dict={"a":1,"b":2,"c":3}
print("Original_dict= ", original_dict)
new_dict={key:value*4 for key,value in original_dict.items()}
print("New_dict= ", new_dict)
```

```python
celsius_temps=[0,10,20,30,40,50]
print("Celsius_temps= ", celsius_temps)
farenheit_temps={str(c)+"c":str((c*9/5)+32)+"f" for c in celsius_temps}
print(farenheit_temps)
```

Sets

-Unordered

-Unindexed

-Mutable

-An object/item cannot appear more than once in a set

-Sets only hold immutable objects

```python
my_set={1,2,3,4,5}
my_set2=set([1,2,3,4,5])
unknown={}
empty_set=set()
print("my_set =",my_set)
print("my_set2 =",my_set2)
print("unknown =",unknown)
print("type(my_set) =",type(my_set))
print("type(my_set2) =",type(my_set2))
print("type(unknown) =",type(unknown))
print("type(empty_set) =",type(empty_set))
print("my_set==my_set2=",my_set==my_set2)

```
